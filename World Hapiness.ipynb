"""
## Synthèse du contrôle qualité

### Structure
- 2 116 observations
- 13 variables
- 168 pays/territoires distincts
- 14 années disponibles

### Granularité
- Une observation correspond à un pays pour une année.
- Le couple Country name + Year est unique.

### Doublons
- Aucun doublon complet.
- Aucun doublon sur Country name + Year.

### Années
- Période couverte : 2011 à 2025.
- L'année 2013 est absente du jeu de données.
- Aucune année ne sera créée artificiellement.

### Valeurs manquantes
- Les variables principales Year, Rank, Country name et Life evaluation
  ne présentent pas de valeurs manquantes.
- Les intervalles de confiance et variables explicatives présentent
  des valeurs manquantes importantes, principalement sur les années anciennes.

### Facteurs explicatifs
- Les facteurs sont principalement disponibles à partir de 2019.
- Quelques valeurs restent manquantes après 2019.
- L'analyse des facteurs devra tenir compte de cette disponibilité.

### Valeurs aberrantes
- Le score Life evaluation est compris entre 1,364 et 7,856.
- Aucune valeur ne se situe en dehors de l'échelle 0–10.

### Points de vigilance
- Certains pays peuvent apparaître sous des noms différents selon les années,
  par exemple Swaziland / Eswatini.
- La colonne Country name devra être normalisée si nécessaire avant
  la création de la dimension géographique
  """

```python
df_2019_2025 = df[df["Year"].between(2019, 2025)]
```

```python
df_2019_2025
```

```python
nb_annees = df["Year"].nunique()
pays_toujours_presents = (
    df.groupby("Country name")["Year"]
      .nunique()
      .loc[lambda x: x == nb_annees]
      .index
)

print(pays_toujours_presents)
```
